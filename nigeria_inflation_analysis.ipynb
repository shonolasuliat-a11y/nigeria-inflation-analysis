```python
import wbgapi as wb

indicators = ["FP.CPI.TOTL.ZG", "PA.NUS.FCRF", "NY.GDP.MKTP.KD.ZG"]
df = wb.data.DataFrame(indicators, "NGA", time=range(1990, 2024))
df = df.transpose()
df = df.rename(columns={
    "FP.CPI.TOTL.ZG": "inflation",
    "NY.GDP.MKTP.KD.ZG": "gdp_growth",
    "PA.NUS.FCRF": "exchange_rate"
})
print(list(df.columns))
```

```python
import matplotlib.pyplot as plt

plt.plot(df.index, df["inflation"], label="inflation")
plt.plot(df.index, df["exchange_rate"], label="exchange_rate")
plt.title("Nigeria: Inflation vs Exchange Rate")
plt.xlabel("Year")
plt.xticks(rotation=90)
plt.legend()
plt.show()
```

```python

fig, ax1 = plt.subplots()

ax1.plot(df.index, df["inflation"], color="blue", label="inflation")
ax1.set_ylabel("Inflation (%)", color="blue")

ax2 = ax1.twinx()
ax2.plot(df.index, df["exchange_rate"], color="red", label="exchange_rate")
ax2.set_ylabel("Exchange rate (₦/$)", color="red")

plt.title("Nigeria: Inflation vs Exchange Rate")
plt.xticks(rotation=90)
plt.show()
```

```python
import statsmodels.api as sm
X = df["exchange_rate" ]
Y= df["inflation" ]
X = sm.add_constant (X)
model = sm.OLS(Y,X).fit()
print(model.summary())
```
